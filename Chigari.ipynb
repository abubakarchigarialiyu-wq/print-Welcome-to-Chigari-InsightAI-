```python
import pandas as pd
import plotly.express as px

from IPython.display import display
from google.colab import files

print("Libraries imported successfully!")
```

```python
data = {
    "State": [
        "Kano", "Lagos", "Kaduna",
        "Kano", "Lagos", "Kaduna"
    ],

    "Year": [
        2019, 2019, 2019,
        2023, 2023, 2023
    ],

    "Registered_Voters": [
        5000000, 6500000, 4000000,
        5800000, 7000000, 4500000
    ],

    "Votes_Cast": [
        2100000, 2200000, 1500000,
        1800000, 1900000, 1300000
    ]
}

df = pd.DataFrame(data)

print("Sample dataset created successfully!")

display(df)
```

```python
df["Turnout_Percent"] = (
    df["Votes_Cast"] /
    df["Registered_Voters"] * 100
).round(2)

print("Turnout calculated successfully!")

display(df)
```

```python
print("DATASET OVERVIEW")

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

print("\nColumn names:")
print(list(df.columns))

print("\nFirst five records:")
display(df.head())
```

```python
print("DATA QUALITY CHECK")

print("\nMissing values:")
display(df.isnull().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())
```

```python
print("STATISTICAL SUMMARY")

display(df.describe().round(2))
```

```python
fig = px.bar(
    df,
    x="State",
    y="Turnout_Percent",
    color="Year",
    barmode="group",
    title="Sample Election Turnout by State and Year",
    labels={
        "State": "State",
        "Turnout_Percent": "Turnout (%)",
        "Year": "Election Year"
    }
)

fig.show()
```

```python
state_summary = (
    df.groupby("State")["Turnout_Percent"]
    .mean()
    .round(2)
    .reset_index()
)


def ask_your_data(question):
    question = question.lower()

    if "highest" in question and "turnout" in question:
        result = state_summary.loc[
            state_summary["Turnout_Percent"].idxmax()
        ]

        return (
            f'{result["State"]} has the highest average '
            f'turnout at {result["Turnout_Percent"]}%.'
        )

    elif "lowest" in question and "turnout" in question:
        result = state_summary.loc[
            state_summary["Turnout_Percent"].idxmin()
        ]

        return (
            f'{result["State"]} has the lowest average '
            f'turnout at {result["Turnout_Percent"]}%.'
        )

    elif "average" in question and "turnout" in question:
        average = round(df["Turnout_Percent"].mean(), 2)

        return f"The average sample turnout is {average}%."

    elif "how many" in question or "number of records" in question:
        return f"The dataset contains {len(df)} records."

    elif "columns" in question:
        return f"Dataset columns: {', '.join(df.columns)}."

    else:
        return (
            "I cannot answer that question yet. Try asking about "
            "the highest turnout, lowest turnout, average turnout, "
            "number of records, or columns."
        )


print(ask_your_data("Which state has the highest turnout?"))
print(ask_your_data("What is the average turnout?"))
print(ask_your_data("How many records are in the dataset?"))
```

```python
# ==========================================
# CHIGARI INSIGHTAI
# SMART DATA ANALYZER - UPGRADE 1
# ==========================================

import pandas as pd
import plotly.express as px
from IPython.display import display, Markdown

# Select the dataset
if "user_df" in globals() and isinstance(user_df, pd.DataFrame):
    analysis_df = user_df.copy()
else:
    analysis_df = df.copy()

print("CHIGARI INSIGHTAI")
print("Turn Data Into Clear Insights")
print("=" * 45)

# Dataset overview
print("\n1. DATASET OVERVIEW")
print("Total records:", len(analysis_df))
print("Total columns:", len(analysis_df.columns))

print("\n2. COLUMN NAMES")
print(list(analysis_df.columns))

# Data quality checks
print("\n3. DATA QUALITY")
print("Missing values:", int(analysis_df.isnull().sum().sum()))
print("Duplicate rows:", int(analysis_df.duplicated().sum()))

# Numerical summary
print("\n4. NUMERICAL SUMMARY")

numeric_df = analysis_df.select_dtypes(include="number")

if not numeric_df.empty:
    display(numeric_df.describe().round(2))
else:
    print("No numerical columns found.")

# Automatically choose a useful chart
print("\n5. AUTOMATIC VISUALIZATION")

if "Turnout_Percent" in analysis_df.columns:
    fig = px.bar(
        analysis_df,
        x="State" if "State" in analysis_df.columns else analysis_df.index,
        y="Turnout_Percent",
        color="Year" if "Year" in analysis_df.columns else None,
        title="Election Turnout Analysis",
        labels={"Turnout_Percent": "Turnout (%)"}
    )

elif len(numeric_df.columns) > 0:
    chart_column = numeric_df.columns[0]

    fig = px.histogram(
        analysis_df,
        x=chart_column,
        title=f"Distribution of {chart_column}"
    )

else:
    fig = None
    print("No suitable numerical column is available for a chart.")

if fig is not None:
    fig.show()

# Generate automatic insights
print("\n6. AUTOMATIC INSIGHTS")

if "Turnout_Percent" in analysis_df.columns and not analysis_df.empty:

    valid_turnout = analysis_df.dropna(subset=["Turnout_Percent"])

    if not valid_turnout.empty:
        highest = valid_turnout.loc[
            valid_turnout["Turnout_Percent"].idxmax()
        ]

        lowest = valid_turnout.loc[
            valid_turnout["Turnout_Percent"].idxmin()
        ]

        average = valid_turnout["Turnout_Percent"].mean()

        print(
            f"Highest turnout: {highest['Turnout_Percent']:.2f}%"
        )

        if "State" in valid_turnout.columns:
            print("State with highest turnout:", highest["State"])

        print(
            f"Lowest turnout: {lowest['Turnout_Percent']:.2f}%"
        )

        if "State" in valid_turnout.columns:
            print("State with lowest turnout:", lowest["State"])

        print(f"Average turnout: {average:.2f}%")

print("\nAnalysis completed successfully!")
print("Note: Sample figures are fictional and are for demonstration only.")
```
